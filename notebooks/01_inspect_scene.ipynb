# Inspect the sequential-grasp scene
This notebook uses offscreen `mujoco.Renderer`. Interactive `mujoco.viewer.launch()` belongs in `seqgrasp/viz/viewer.py` and must not be called here.

```python
import mediapy as media
import mujoco
from seqgrasp import load_configs
from seqgrasp.scene_builder import build_scene
model, data = build_scene(load_configs())
frames = []
with mujoco.Renderer(model) as renderer:
    for _ in range(3):
        mujoco.mj_step(model, data)
        renderer.update_scene(data)
        frames.append(renderer.render())
media.show_images(frames)
```
